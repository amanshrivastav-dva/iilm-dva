```python
# ==========================================
# Using NumPy for Descriptive Statistics
# Calculate Mean, Median, Standard Deviation,
# and Variance of the 'charges' column
# ==========================================

import numpy as np

# Mean
print("Mean of Charges:", np.mean(df["charges"]))

# Median
print("Median of Charges:", np.median(df["charges"]))

# Standard Deviation
print("Standard Deviation of Charges:", np.std(df["charges"], ddof=1))

# Variance
print("Variance of Charges:", np.var(df["charges"], ddof=1))
```

```python
# ==========================================
# Using SciPy for Descriptive Statistics
# Calculate Mode of the 'children' column
# ==========================================

from scipy import stats

# Calculate Mode
mode_result = stats.mode(df["children"])

print("Mode of Children:", mode_result.mode[0])
print("Frequency:", mode_result.count[0])
```

```python
# Display all statistical measures together
print("Mean:\n", df.mean(numeric_only=True))
print("\nMedian:\n", df.median(numeric_only=True))
print("\nMode:\n", df.mode().iloc[0])
print("\nVariance:\n", df.var(numeric_only=True))
print("\nStandard Deviation:\n", df.std(numeric_only=True))
print("\nQuartiles:\n", df.quantile([0.25,0.5,0.75], numeric_only=True))
print("\nCorrelation:\n", df.corr(numeric_only=True))
```
